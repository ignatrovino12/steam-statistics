```python
import json
import pandas as pd

INPUT_PATH = "../../data/raw/steam_apps.json"

with open(INPUT_PATH, "r", encoding="utf-8") as file:
    data = json.load(file)

apps = data["response"]["apps"]

df = pd.DataFrame(apps)

print(df.head())
print()
print(df.shape)
print()
print(df.info())
```
